```python
# Loading the downloaded dataset from local environment
df = pd.read_csv(r"../data/Sample - Superstore.csv", encoding='latin-1')
```

```python
df.head()
```

```python
# 3. DATA EXPLORATION

# 1️ First 5 rows
print("First 5 Rows:")
print(df.head())

# 2️ Dataset shape
print("\nDataset Shape (rows, columns):", df.shape)

# 3️ Column names
print("\nColumns:")
print(df.columns.tolist())

# 4️ Data types & null values
print("\nDataset Info:")
df.info()

# 5️ Summary statistics (numeric only)
print("\nNumeric Summary Statistics:")
print(df.describe())

# 6️ Summary statistics (all columns including categorical)
print("\nFull Summary Statistics (all columns):")
print(df.describe(include='all'))

# 7️ Count of missing values per column
print("\nMissing Values per Column:")
print(df.isnull().sum())
```

```python
# 8️ Check for duplicate rows
duplicates = df.duplicated().sum()
print("\nNumber of Duplicate Rows:", duplicates)

# 9️ Categorical column value counts
categorical_columns = df.select_dtypes(include='object').columns
print("\nCategorical Column Value Counts (top 5 each):")
for col in categorical_columns:
    print(f"\nColumn: {col}")
    print(df[col].value_counts().head())

# 10 Correlation matrix
numeric_df = df.select_dtypes(include='number')
corr_matrix = numeric_df.corr()

print("\nCorrelation Matrix (numeric columns only):")
print(corr_matrix)

# Correlation heatmap
import seaborn as sns
import matplotlib.pyplot as plt

# Visualize heatmap
plt.figure(figsize=(8, 6))
sns.heatmap(corr_matrix, annot=True, cmap='coolwarm', fmt=".2f")
plt.title("Correlation Heatmap (Numeric Columns)")
plt.show()
```

```python
# 4. TOTAL STORE REVENUE
total_sales = df["Sales"].sum()

print("\nTotal Revenue:")
print(total_sales)
```

```python
# 5. WHICH REGION GENERATES THE MOST REVENUE?
region_sales = df.groupby("Region")["Sales"].sum()

print("\nTotal Sales by Region:")
print(region_sales)

# Sort regions by sales
region_sales_sorted = region_sales.sort_values(ascending=False)

print("\nRegions Sorted by Sales:")
print(region_sales_sorted)
```

```python
# 6. VISUALIZATION: SALES BY REGION
region_sales.plot(kind="bar")

plt.title("Sales by Region")
plt.xlabel("Region")
plt.ylabel("Total Sales")

plt.show()
```

```python
# 7. WHAT PRODUCT CATEGORY SELLS THE MOST?
category_sales = df.groupby("Category")["Sales"].sum()

print("\nSales by Category:")
print(category_sales)

# Identify top selling category
top_category = category_sales.idxmax()

print("\nTop Selling Category:")
print(top_category)
```

```python
# 8. VISUALIZATION: SALES BY CATEGORY
category_sales.plot(kind="bar")

plt.title("Sales by Category")
plt.xlabel("Category")
plt.ylabel("Total Sales")

plt.show()
```

```python
# 9. WHICH REGION HAS THE MOST ORDERS?
region_orders = df["Region"].value_counts()

print("\nNumber of Orders by Region:")
print(region_orders)
```

```python
# 10. TOP 5 HIGHEST SALES ORDERS
top_sales = df.sort_values(by="Sales", ascending=False)

print("\nTop 5 Highest Sales Orders:")
print(top_sales.head(5))
```

```python
# 11. FILTERING EXAMPLE – WEST REGION ORDERS
west_orders = df[df["Region"] == "West"]

print("\nOrders from the West Region:")
print(west_orders.head())
```

```python
# 12. FILTERING EXAMPLE – HIGH VALUE ORDERS
# Orders where sales are greater than $1000
high_sales_orders = df[df["Sales"] > 1000]

print("\nOrders with Sales Greater Than $1000:")
print(high_sales_orders.head())
```

```python
# 13. AVERAGE SALES PER CATEGORY
avg_sales_category = df.groupby("Category")["Sales"].mean()

print("\nAverage Sales by Category:")
print(avg_sales_category)
```

```python
# 14. MONTHLY SALES TREND
# Convert Order Date to datetime
df["Order Date"] = pd.to_datetime(df["Order Date"])

# Extract month
df["Month"] = df["Order Date"].dt.to_period("M")

# Calculate monthly sales
monthly_sales = df.groupby("Month")["Sales"].sum()

print("\nMonthly Sales:")
print(monthly_sales.head())
```

```python
# 15. VISUALIZATION: MONTHLY SALES TREND
monthly_sales.plot()

plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Total Sales")

plt.show()
```

```python
# 16. MOST PROFITABLE CATEGORY

category_profit = df.groupby("Category")["Profit"].sum()

print("\nProfit by Category:")
print(category_profit)

top_category_profit = category_profit.idxmax()

print("\nMost Profitable Category:")
print(top_category_profit)
```

```python
# 17. MOST PROFITABLE REGION

region_profit = df.groupby("Region")["Profit"].sum()

print("\nProfit by Region:")
print(region_profit)

top_region_profit = region_profit.idxmax()

print("\nMost Profitable Region:")
print(top_region_profit)
```

```python
# Visualize total profit by region
region_profit.plot(kind="bar")

plt.title("Total Profit by Region")
plt.xlabel("Region")
plt.ylabel("Total Profit")

plt.show()
```
