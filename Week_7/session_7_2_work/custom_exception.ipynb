## Custom Exception Example

```python
# Define custom exceptions

class InvalidDataError(Exception):
    """Raised when data validation fails"""
    pass

class ProcessingError(Exception):
    """Raised during data processing"""
    pass
```

```python
# Use custom exceptions

def process_data(data):
    if data is None:
        raise InvalidDataError('Data cannot be None')
    if len(data) == 0:
        raise InvalidDataError('Data cannot be empty')
    # Process...
    processed_data = [item * 2 for item in data]
    return processed_data
```

```python
# Clear, specific error types!
```

```python
# Usage:

data = None
result = process_data(data)  # Pass the 'data' variable as an argument to the function
print(result)  # This will print the processed data
```

```python
data = ()
result = process_data(data)  # Pass the 'data' variable as an argument to the function
print(result)  # This will print the processed data
```

```python
data = (1,2,3)
result = process_data(data)  # Pass the 'data' variable as an argument to the function
print(result)  # This will print the processed data
```
